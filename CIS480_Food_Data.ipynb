```python
#whether poverty correlates with low‑income food access barriers.
plt.scatter(df_cook_selected['PovertyRate'], df_cook_selected['lalowi1share'])
plt.title('Low-Income Food Access vs Poverty – Cook County, Illinois')
plt.xlabel('Poverty Rate (%)')
plt.ylabel('Low-Income Population >1 Mile From Supermarket (%)')
plt.grid(True)
plt.show()
```

```python
#households without vehicles are concentrated in high‑poverty areas.
plt.scatter(df_cook_selected['PovertyRate'], df_cook_selected['lahunv1share'])
plt.title('Vehicle Access Vulnerability vs Poverty – Cook County, Illinois')
plt.xlabel('Poverty Rate (%)')
plt.ylabel('Households Without Vehicles >1 Mile From Supermarket (%)')
plt.grid(True)
plt.show()
```

```python
#SNAP households are disproportionately located in areas with poor food access.
plt.scatter(df_cook_selected['lasnap1share'], df_cook_selected['lapop1share'])
plt.title('SNAP Reliance vs Food Access – Cook County, Illinois')
plt.xlabel('SNAP Households >1 Mile From Supermarket (%)')
plt.ylabel('Population >1 Mile From Supermarket (%)')
plt.grid(True)
plt.show()
```
